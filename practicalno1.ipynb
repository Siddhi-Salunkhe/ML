```python
import numpy as np
```

```python
X=np.array([1,2,3,4,5])
```

```python
Y=np.array([70,80,90,95,85])
```

```python
x_mean=np.mean(X)
```

```python
y_mean=np.mean(Y)
```

```python
m=np.sum((X-x_mean)*(Y-y_mean))/np.sum((X-x_mean)**2)
```

```python
c=y_mean-m*x_mean
```

```python
Y_pred=m*X+c
```

```python
print("Slope m =", m)
```

```python
print("Intercept c =", c)
```

```python
print("Equation of the line: Y =", m, "X +", c)
```
